```python
import os
import re
import unicodedata

def remove_vietnamese_accents(text):
    if not text:
        return ""
    text = unicodedata.normalize('NFC', text)
    patterns = {
        '[àáảãạăắằẳẵặâấầẩẫậ]': 'a', '[ÀÁẢÃẠĂẮẰẲẴẶÂẤẦẨẪẬ]': 'A',
        '[èéẻẽẹêếềểễệ]': 'e', '[ÈÉÈẼẸÊẾỀỂỄỆ]': 'E',
        '[ìíỉĩị]': 'i', '[ÌÍỈĨỊ]': 'I',
        '[òóỏõọôốồổỗộơớờởỡợ]': 'o', '[ÒÓỎÕỌÔỐỒỔỖỘƠỚỜỞỠỢ]': 'O',
        '[ùúủũụưứừửữự]': 'u', '[ÙÚỦŨỤƯỨỪỬỮỰ]': 'U',
        '[ỳýỷỹỵ]': 'y', '[ỲÝỶỸỴ]': 'Y',
        '[đ]': 'd', '[Đ]': 'D'
    }
    for pattern, replacement in patterns.items():
        text = re.sub(pattern, replacement, text)
    return text

def clean_name(name):
    # Loại bỏ dấu tiếng Việt
    name = remove_vietnamese_accents(name)
    # Thay khoảng trắng bằng gạch dưới
    name = name.replace(" ", "_")
    # Loại bỏ ký tự đặc biệt trừ gạch dưới, gạch ngang và chữ số
    name = re.sub(r'[^\w\s-]', '', name)
    # Thay thế nhiều dấu gạch dưới liên tiếp bằng 1 cái cho đẹp
    name = re.sub(r'_+', '_', name)
    return name

def rename_everything(root_folder):
    # topdown=False: Quan trọng! Đổi tên file/thư mục con trước, 
    # sau đó mới đổi tên thư mục cha. Nếu đổi tên cha trước, 
    # đường dẫn con sẽ bị hỏng.
    for root, dirs, files in os.walk(root_folder, topdown=False):
        
        # 1. Đổi tên các file
        for filename in files:
            old_path = os.path.join(root, filename)
            name_part, extension = os.path.splitext(filename)
            new_filename = clean_name(name_part) + extension
            new_path = os.path.join(root, new_filename)
            
            if old_path != new_path:
                if os.path.exists(new_path):
                    new_path = os.path.join(root, clean_name(name_part) + "_1" + extension)
                os.rename(old_path, new_path)
                print(f"File: {filename} -> {new_filename}")

        # 2. Đổi tên các thư mục con
        for dirname in dirs:
            old_path = os.path.join(root, dirname)
            new_dirname = clean_name(dirname)
            new_path = os.path.join(root, new_dirname)
            
            if old_path != new_path:
                if os.path.exists(new_path):
                    new_path = os.path.join(root, new_dirname + "_dir")
                os.rename(old_path, new_path)
                print(f"Folder: {dirname} -> {new_dirname}")

# Đường dẫn gốc của bạn
root_path = r"D:\RAG\UET_Database_Cleaned"
 
if __name__ == "__main__":
    rename_everything(root_path)
    print("\n--- XONG! Toàn bộ file và thư mục đã được làm sạch ---")
```
